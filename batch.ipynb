```python
import pandas as pd
import numpy as np

from stream_analysis.sites.youtube import get_playlist_videos
from stream_analysis.utils import minutes_to_hhmm
from stream_analysis.chat import Chat

import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
video_envs = get_playlist_videos('PLECPddeTU9ojxLzMI7U2VybeBO_pvOIpX')
video_data = {
    'Liver': [],
    'StartAt': [],
    'EndAt': [],
    'Duration': [],
    'Title': [],
}

for env_ in video_envs:
    video_data['Liver'].append(env_.video_liver)
    video_data['Title'].append(env_.video_title)
    video_data['StartAt'].append(np.datetime64(env_.video_start_at, 's'))
    video_data['EndAt'].append(np.datetime64(env_.video_end_at, 's'))
    video_data['Duration'].append(minutes_to_hhmm(env_.video_data.get('duration') / 60))

video_data = pd.DataFrame(video_data)
video_data
```
